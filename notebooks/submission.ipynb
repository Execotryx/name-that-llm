# Name That LLM offline submission
Attach the trusted release bundle and matching dependency lock. Set NTL_BUNDLE and measured NTL_AVAILABLE_MEMORY_BYTES. Restart with Internet disabled and set NTL_COLD_RESTART=1. This notebook performs fixed inference only.


```python
import os
import sys
from pathlib import Path
bundle = Path(os.environ['NTL_BUNDLE'])
sys.path.insert(0, str(bundle / 'source'))
from ntl.release import replay
report = replay(
    bundle,
    Path(os.environ.get('NTL_TEST', '/kaggle/input/competitions/name-that-llm/test.csv')),
    Path('/kaggle/working'),
    available_memory=int(os.environ['NTL_AVAILABLE_MEMORY_BYTES']),
    cold_restart=os.environ.get('NTL_COLD_RESTART') == '1',
)
print({'status': report['status'], 'rows': report['rows'], 'wall_seconds': report['wall_seconds']})
```
